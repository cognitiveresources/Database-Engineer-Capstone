```python
import mysql.connector as connector
connection = connector.connect(user = "root", password = "root", database = "LittleLemonDB")
# mysql querry stored in a variable
promo_query = """
SELECT Customers.CustomerID, Customers.FullName, Customers.ContactNumber, Customers.Email, Orders.TotalCost
FROM Customers
RIGHT JOIN  Orders ON Customers.CustomerID = Orders.CustomerID
WHERE Orders.TotalCost > 60
"""
# execute querry via cursor object
cursor = connection.cursor(buffered = True)
cursor.execute(promo_query)
# fetch results
results = cursor.fetchall()
# print results
print("Customer ID, Full Name, Phone number, Email, Bill")
for row in results:
    print(row)
# close connection
cursor.close()
connection.close()
```
